##### **Silver Layer**
- ##### *Clean & standardize raw (dbo) tables → silver*

```sql
--  Create silver schema

CREATE SCHEMA IF NOT EXISTS silver;
```

```sql
-- Validate source tables in dbo

SHOW TABLES IN dbo;
```

```sql
-- Load raw tables into silver layer (1:1 structure)

CREATE OR REPLACE TABLE silver.channels AS
SELECT * FROM dbo.channels_raw;

CREATE OR REPLACE TABLE silver.customer_addresses AS
SELECT * FROM dbo.customer_addresses_raw;

CREATE OR REPLACE TABLE silver.customer_segments AS
SELECT * FROM dbo.customer_segments_raw;

CREATE OR REPLACE TABLE silver.customers AS
SELECT * FROM dbo.customers_raw;

CREATE OR REPLACE TABLE silver.geography AS
SELECT * FROM dbo.geography_raw;

CREATE OR REPLACE TABLE silver.order_items AS
SELECT * FROM dbo.order_items_raw;

CREATE OR REPLACE TABLE silver.order_statuses AS
SELECT * FROM dbo.order_statuses_raw;

CREATE OR REPLACE TABLE silver.orders AS
SELECT * FROM dbo.orders_raw;

CREATE OR REPLACE TABLE silver.products AS
SELECT * FROM dbo.products_raw;

CREATE OR REPLACE TABLE silver.region AS
SELECT * FROM dbo.region_raw;
```

```sql
-- Validate row counts across all silver tables

SELECT 'channels' AS table_name, COUNT(*) AS row_count FROM silver.channels
UNION ALL
SELECT 'customer_segments', COUNT(*) FROM silver.customer_segments
UNION ALL
SELECT 'region', COUNT(*) FROM silver.region
UNION ALL
SELECT 'geography', COUNT(*) FROM silver.geography
UNION ALL
SELECT 'customers', COUNT(*) FROM silver.customers
UNION ALL
SELECT 'customer_addresses', COUNT(*) FROM silver.customer_addresses
UNION ALL
SELECT 'products', COUNT(*) FROM silver.products
UNION ALL
SELECT 'order_statuses', COUNT(*) FROM silver.order_statuses
UNION ALL
SELECT 'orders', COUNT(*) FROM silver.orders
UNION ALL
SELECT 'order_items', COUNT(*) FROM silver.order_items;
```
